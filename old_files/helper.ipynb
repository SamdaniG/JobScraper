```python
from dataclasses import dataclass, asdict
from typing import Optional
class BaseModel:
    def to_dict(self):
        return {
            k: v
            for k, v in asdict(self).items()
            if v is not None
        }

@dataclass(kw_only=True)
class Job(BaseModel):
    job_id: str
    job_name: str
    source: str
    work_policy: Optional[str] = None
    location: str
    posted_date: str
    filled_date: Optional[str] = ''
    url: str

```
