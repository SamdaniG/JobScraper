```python
import json

# Load the DB
with open("../db.json", "r") as f:
    db = json.load(f)

# Collect keys to delete
to_delete = []

for hash_id, job in db.items():
    if (
        job.get("source") == "rivian" and
        job.get("filled_date", "") == ""
    ):
        to_delete.append(hash_id)

# Delete them
for hash_id in to_delete:
    del db[hash_id]

print(f"Removed {len(to_delete)} jobs.")

# Write updated DB back
with open("../db.json", "w") as f:
    json.dump(db, f, indent=4)
```
