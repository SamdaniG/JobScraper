```python
import json

# Load the DB
with open("../db.json", "r") as f:
    db = json.load(f)

# Collect keys to delete
to_delete = []

for hash_id, job in db.items():
    if (
        job.get("source") == "kepler" and
        job.get("filled_date", "") == ""
    ):
        to_delete.append(hash_id)

# Delete them
for hash_id in to_delete:
    del db[hash_id]

print(f"Removed {len(to_delete)} jobs.")

# Write updated DB back
with open("../db.json", "w") as f:
    json.dump(db, f, indent=4)
```

CLeaning the db, so that hiring manger becomes hiring manager

```python
import json

# Load the DB
with open("../db.json", "r") as f:
    db = json.load(f)



for hash_id, job_detail in db.items():
    if job_detail.get("hiring_manger", None) is not None:
        name=job_detail['hiring_manger']
        job_detail.pop("hiring_manger")
        job_detail['hiring_manager']=name
        # print(hash_id)
        print(job_detail)
```

```python
with open("../db.json", "w") as f:
    json.dump(db, f, indent=4)
```
