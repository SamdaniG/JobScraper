```python
import requests

url = "https://careers.powerco.de/services/jobs/options/facetValues/"

payload = {
    "page": 0,
    "keywords": "",
    "locationsearch": "",
    "sortby": "referencedate",
    "sortdir": "desc",
    "sortfield": "title",
    "startrow": 0,
    "recordsperpage": 25,
    "facetquery": {
        "facet": True,
        "mincount": 1,
        "limit": 5000
    },
    "filterquery": {}
}

headers = {
    "Content-Type": "application/json"
}

resp = requests.post(url, json=payload, headers=headers)
print(resp.status_code)
print(resp.json())
```

```python
# models/job.py
from dataclasses import dataclass, asdict
from typing import Optional

@dataclass(kw_only=True)
class Job:
    job_id: str
    job_name: str
    source: str
    work_policy: Optional[str] = None
    location: str
    posted_date: str
    filled_date: Optional[str] = None
    url: str


    def to_dict(self):
        return asdict(self)
```
