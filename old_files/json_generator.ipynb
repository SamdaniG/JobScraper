```python
FMT='%Y-%m-%dT%H:%M:%Sz'
DATE_FMT = "%a %d-%b-%Y"
```

```python
import sqlite3
import json
from datetime import datetime

DB_PATH = "jobs.db"
JSON_PATH = "new_db.json"

conn = sqlite3.connect(DB_PATH)
conn.row_factory = sqlite3.Row
cursor = conn.cursor()

cursor.execute("SELECT * FROM jobs")
rows = cursor.fetchall()

jobs = {}

for row in rows:
    row = dict(row)

    # Use hash_id as the JSON key
    hash_id = row.pop("hash_id")

    # Remove any fields with NULL values
    row = {k: v for k, v in row.items() if v is not None}

    if row.get("posted_date"):
        row["posted_date"]=datetime.strptime(row["posted_date"],FMT).strftime(DATE_FMT)

    if row.get("updated_date"):
        row["updated_date"]=datetime.strptime(row["updated_date"],FMT).strftime(DATE_FMT)

    if row.get("filled_date"):
        row["filled_date"]=datetime.strptime(row["filled_date"],FMT).strftime(DATE_FMT)

    if row.get("creation_date"):
        row["creation_date"]=datetime.strptime(row["creation_date"],FMT).strftime(DATE_FMT)

    jobs[hash_id] = row

with open(JSON_PATH, "w") as f:
    json.dump(jobs, f, indent=4)

conn.close()

print(f"Exported {len(jobs)} jobs to {JSON_PATH}")
```
